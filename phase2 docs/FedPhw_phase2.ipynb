```python
def process_chunk1(X, y, start_idx, end_idx, seq_len):
    X_seq = []
    y_seq = []

    for i in range(start_idx, end_idx - seq_len + 1):
        X_seq.append(X[i:i + seq_len])
        y_seq.append(y[i + seq_len - 1])

    return np.array(X_seq, dtype=np.float32), np.array(y_seq, dtype=np.int64)

def process_chunk(X, y, start_idx, end_idx, seq_len, stride=5):
    X_seq = []
    y_seq = []

    for i in range(start_idx, end_idx - seq_len + 1, stride):
        X_seq.append(X[i:i + seq_len])
        y_seq.append(y[i + seq_len - 1])

    return np.array(X_seq, dtype=np.float32), np.array(y_seq, dtype=np.int64)
```

```python
import random

def iid_partition(num_chunks, clients):
    chunk_ids = list(range(num_chunks))
    random.shuffle(chunk_ids)

    assignments = {c: [] for c in clients}
    for idx, cid in enumerate(chunk_ids):
        assignments[clients[idx % len(clients)]].append(cid)

    return assignments

def non_iid_label_skew_partition(y_chunks, clients, dominance=0.7):

    client_assignments = {c: [] for c in clients}

    # analyze chunk labels
    attack_chunks = []
    benign_chunks = []

    for i, y_file in enumerate(y_chunks):
        y = np.load(os.path.join(SEQ_CHUNK_DIR, y_file), mmap_mode="r")
        attack_ratio = y.mean()  # since attack=1, benign=0

        if attack_ratio > 0.5:
            attack_chunks.append(i)
        else:
            benign_chunks.append(i)

    random.shuffle(attack_chunks)
    random.shuffle(benign_chunks)

    for idx, client in enumerate(clients):
        num_attack = int(dominance * len(attack_chunks) / len(clients))
        num_benign = int((1 - dominance) * len(benign_chunks) / len(clients))

        client_assignments[client].extend(attack_chunks[:num_attack])
        client_assignments[client].extend(benign_chunks[:num_benign])

        del attack_chunks[:num_attack]
        del benign_chunks[:num_benign]

    return client_assignments
if DATA_DISTRIBUTION == "iid":
    client_assignments = iid_partition(NUM_CHUNKS, CLIENTS)

elif DATA_DISTRIBUTION == "non_iid_label_skew":
    client_assignments = non_iid_label_skew_partition(y_chunks, CLIENTS)

else:
    raise ValueError("Unknown DATA_DISTRIBUTION")
```

```python
class CNN_LSTM_IDS(nn.Module):
    def __init__(self, seq_len, num_features):
        super().__init__()

        # Spatial feature extraction
        self.conv1 = nn.Conv1d(
            in_channels=num_features,
            out_channels=64,
            kernel_size=3,
            padding=1
        )
        self.relu = nn.ReLU()

        # Temporal modeling
        self.lstm = nn.LSTM(
            input_size=64,
            hidden_size=64,
            num_layers=1,
            batch_first=True
        )

        # Classifier
        self.fc = nn.Linear(64, 1)
        #self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # x: (B, T, F) -> Conv1d expects (B, F, T)
        x = x.permute(0, 2, 1)
        x = self.relu(self.conv1(x))
        # back to (B, T, C)
        x = x.permute(0, 2, 1)

        _, (h_n, _) = self.lstm(x)
        h_last = h_n[-1]           # (B, 64)
        out = self.fc(h_last)
        return out
```

```python
from torch.utils.data import Dataset
import numpy as np
import os
import torch

class ClientSequenceDataset(Dataset):
    def __init__(self, client_dir):
        self.x_files = sorted([
            os.path.join(client_dir, f)
            for f in os.listdir(client_dir) if f.startswith("X_seq")
        ])
        self.y_files = sorted([
            os.path.join(client_dir, f)
            for f in os.listdir(client_dir) if f.startswith("y_seq")
        ])

        assert len(self.x_files) == len(self.y_files)

        self.chunk_sizes = []
        for yf in self.y_files:
            y = np.load(yf, mmap_mode="r")
            self.chunk_sizes.append(len(y))

        self.cumulative_sizes = np.cumsum(self.chunk_sizes)


        self._current_chunk_id = None
        self._current_x = None
        self._current_y = None

    def __len__(self):
        return int(self.cumulative_sizes[-1])

    def __getitem__(self, idx):
        chunk_id = np.searchsorted(self.cumulative_sizes, idx, side="right")
        local_idx = idx if chunk_id == 0 else idx - self.cumulative_sizes[chunk_id - 1]


        if chunk_id != self._current_chunk_id:
            self._current_x = np.load(self.x_files[chunk_id], mmap_mode="r")
            self._current_y = np.load(self.y_files[chunk_id], mmap_mode="r")
            self._current_chunk_id = chunk_id

        x = self._current_x[local_idx]
        y = self._current_y[local_idx]

        return (
            torch.tensor(x, dtype=torch.float32),
            torch.tensor(y, dtype=torch.float32)
        )

```

```python
MAX_BATCHES = 50

def local_train(model, dataloader, epochs, lr):
    model.train()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = CRITERION

    for epoch in range(epochs):
        for batch_idx, (x, y) in enumerate(dataloader):
            if batch_idx >= MAX_BATCHES:
                break

            x = x.to(DEVICE)
            y = y.to(DEVICE).unsqueeze(1)

            optimizer.zero_grad()
            preds = model(x)
            loss = criterion(preds, y)
            loss.backward()
            optimizer.step()

    return model.state_dict()



def fedavg(state_dicts, data_sizes):
    avg_state = {}
    total = sum(data_sizes)

    for key in state_dicts[0].keys():
        avg_state[key] = sum(
            state_dicts[i][key].float() * data_sizes[i] / total
            for i in range(len(state_dicts))
        )

    return avg_state

```

```python

import tenseal as ts
import numpy as np
import torch


HE_POLY_MODULUS = 16384
HE_SCALE_BITS = 40
HE_COEFF_MOD_BITS = [60, 40, 40, 40, 40, 60]

# Selected layers to encrypt (60% faster)
SELECTED_LAYERS = [
    "lstm.weight_ih_l0",
    "lstm.weight_hh_l0",
    "fc.weight",
    "fc.bias"
]


def create_ckks_context():
    """Creates CKKS context - call ONCE before training"""
    context = ts.context(
        ts.SCHEME_TYPE.CKKS,
        poly_modulus_degree=HE_POLY_MODULUS,
        coeff_mod_bit_sizes=HE_COEFF_MOD_BITS
    )
    context.global_scale = 2 ** HE_SCALE_BITS
    context.generate_galois_keys()
    return context
def compute_model_update(local_model, global_model):
    """Computes ΔW = W_local - W_global"""
    delta = {}
    local_state = local_model.state_dict()
    global_state = global_model.state_dict()

    for key in SELECTED_LAYERS:
        if key in local_state and key in global_state:
            diff = local_state[key] - global_state[key]
            # Clip extreme values
            diff = torch.clamp(diff, min=-10.0, max=10.0)
            delta[key] = diff

    return delta


def encrypt_update(delta, context):
    """Encrypts weight delta"""
    encrypted = {}
    shapes = {}

    for key, tensor in delta.items():
        shapes[key] = tensor.shape
        flat = tensor.cpu().detach().numpy().flatten()

        # Validate
        if np.isnan(flat).any() or np.isinf(flat).any():
            flat = np.nan_to_num(flat, nan=0.0, posinf=0.0, neginf=0.0)

        # Clip
        flat = np.clip(flat, -10.0, 10.0)

        # Encrypt
        encrypted[key] = ts.ckks_vector(context, flat.tolist())

    return encrypted, shapes


def encrypted_sum(encrypted_updates):
    """Sums encrypted updates from all clients"""
    if not encrypted_updates:
        return {}

    result = {}
    all_keys = encrypted_updates[0].keys()

    for key in all_keys:
        # Start with first client
        result[key] = encrypted_updates[0][key]

        # Add remaining clients
        for i in range(1, len(encrypted_updates)):
            result[key] = result[key] + encrypted_updates[i][key]

    return result

def decrypt_update(encrypted_sum, shapes):
    """Decrypts aggregated update"""
    decrypted = {}

    for key, enc_vec in encrypted_sum.items():
        # Decrypt
        flat = enc_vec.decrypt()
        flat = np.array(flat, dtype=np.float32)

        # Validate
        flat = np.nan_to_num(flat, nan=0.0, posinf=0.0, neginf=0.0)

        # Reshape
        shape = shapes[key]
        num_elements = np.prod(shape)
        flat = flat[:num_elements]

        tensor = torch.tensor(flat, dtype=torch.float32)
        tensor = tensor.reshape(shape)

        decrypted[key] = tensor

    return decrypted
```

```python
import os
import torch

CHECKPOINT_DIR = os.path.join(BASE_DIR, "fedphe_checkpoints2")
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

def save_checkpoint(round_idx, model):
    ckpt_path = os.path.join(CHECKPOINT_DIR, f"fedphe_round_{round_idx}.pt")
    torch.save({
        "round": round_idx,
        "model_state": model.state_dict()
    }, ckpt_path)
    print(f" Checkpoint saved: {ckpt_path}")


def load_latest_checkpoint(model):
    if not os.path.exists(CHECKPOINT_DIR):
        return 0  # start from scratch

    ckpts = [f for f in os.listdir(CHECKPOINT_DIR) if f.endswith(".pt")]
    if not ckpts:
        return 0

    ckpts.sort(key=lambda x: int(x.split("_")[-1].split(".")[0]))
    latest_ckpt = ckpts[-1]

    ckpt_path = os.path.join(CHECKPOINT_DIR, latest_ckpt)
    data = torch.load(ckpt_path, map_location=DEVICE)

    model.load_state_dict(data["model_state"])
    start_round = data["round"] + 1

    print(f" Resuming from checkpoint: {ckpt_path}")
    print(f"  Starting from round {start_round+1}")

    return start_round
```

```python
import tenseal as ts
import numpy as np
import torch
import torch.nn as nn


HE_POLY_MODULUS = 16384
HE_SCALE_BITS = 40
HE_COEFF_MOD_BITS = [60, 40, 40, 40, 40, 60]

SELECTED_LAYERS = [
    "lstm.weight_ih_l0",
    "lstm.weight_hh_l0",
    "fc.weight",
    "fc.bias"
]

class CNN_LSTM_IDS(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(input_size=8, hidden_size=32, batch_first=True)
        self.fc = nn.Linear(32, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])

def create_ckks_context():
    context = ts.context(
        ts.SCHEME_TYPE.CKKS,
        poly_modulus_degree=HE_POLY_MODULUS,
        coeff_mod_bit_sizes=HE_COEFF_MOD_BITS
    )
    context.global_scale = 2 ** HE_SCALE_BITS
    context.generate_galois_keys()
    context.generate_relin_keys()
    return context

def compute_model_update(local_model, global_model):
    delta = {}
    global_state = global_model.state_dict()
    local_state = local_model.state_dict()

    for key in SELECTED_LAYERS:
        delta[key] = local_state[key] - global_state[key]

    return delta


def encrypt_update(delta, context):
    encrypted = {}
    shapes = {}

    for key, tensor in delta.items():
        flat = tensor.flatten().detach().numpy()
        shapes[key] = tensor.shape
        encrypted[key] = ts.ckks_vector(context, flat.tolist())

    return encrypted, shapes


def encrypted_sum(list_of_updates):
    agg = {}

    for key in list_of_updates[0].keys():
        agg[key] = list_of_updates[0][key]
        for upd in list_of_updates[1:]:
            agg[key] += upd[key]

    return agg


def decrypt_update(enc_update, shapes):
    decrypted = {}

    for key, enc_vec in enc_update.items():
        plain = np.array(enc_vec.decrypt())
        decrypted[key] = torch.tensor(
            plain.reshape(shapes[key]),
            dtype=torch.float32
        )

    return decrypted



def run_weighted_he_demo():

    torch.manual_seed(42)
    np.random.seed(42)

    CLIENT_DATA_SIZES = {
        "Bank_A": 160_000,
        "Bank_B": 160_000,
        "Bank_C":  82_443,
    }

    CLIENTS = list(CLIENT_DATA_SIZES.keys())
    total_data = sum(CLIENT_DATA_SIZES.values())

    client_weights = {
        c: CLIENT_DATA_SIZES[c] / total_data
        for c in CLIENTS
    }

    print("\nClient Weights:")
    for c in CLIENTS:
        print(f"{c}: {client_weights[c]:.6f}")

    # Create CKKS context
    ckks_ctx = create_ckks_context()

    print("\nCKKS Parameters:")
    print("Poly modulus degree:", HE_POLY_MODULUS)
    print("Coeff mod chain:", HE_COEFF_MOD_BITS)
    print("Scale:", f"2^{HE_SCALE_BITS}")

    global_model = CNN_LSTM_IDS()

    # Simulate local training
    local_models = {}
    for c in CLIENTS:
        m = CNN_LSTM_IDS()
        m.load_state_dict(global_model.state_dict())
        with torch.no_grad():
            for p in m.parameters():
                p.data += torch.randn_like(p) * 0.05
        local_models[c] = m

    # Compute weighted deltas
    all_enc_updates = []
    last_shapes = {}

    for c in CLIENTS:
        delta = compute_model_update(local_models[c], global_model)

        weighted_delta = {
            k: v * client_weights[c]
            for k, v in delta.items()
        }

        enc_delta, shapes = encrypt_update(weighted_delta, ckks_ctx)
        all_enc_updates.append(enc_delta)
        last_shapes = shapes

        print(f"\nEncrypted update from {c} (weight={client_weights[c]:.6f})")

        for key, enc_vec in enc_delta.items():
            print(f"Layer: {key}")
            print("  Plain shape:", shapes[key])
            print("  Ciphertext type:", type(enc_vec))
            slot_count = int(np.prod(shapes[key]))
            print("  Packed slots:", slot_count)

    enc_agg = encrypted_sum(all_enc_updates)

    delta_avg = decrypt_update(enc_agg, last_shapes)

    print("\nDecrypted :")
    for key, tensor in delta_avg.items():
        print(f"{key}: {tensor.flatten()[:5].tolist()}")

    # Apply update
    print("\nModel Parameter Update :")
    before = global_model.state_dict()["fc.bias"].clone()

    new_state = global_model.state_dict()
    for key in delta_avg:
        new_state[key] = new_state[key] + delta_avg[key]

    global_model.load_state_dict(new_state)
    after = global_model.state_dict()["fc.bias"]

    print("Before:", before.tolist())
    print("After :", after.tolist())


run_weighted_he_demo()
```

```python
def evaluate_model(model, dataloader, threshold=0.3):
    y_true, y_pred, y_prob = [], [], []

    model.eval()
    with torch.no_grad():
        for x, y in dataloader:
            x = x.to(DEVICE)
            y = y.cpu().numpy()

            logits = model(x)
            probs = torch.sigmoid(logits).cpu().numpy().flatten()
            preds = (probs > threshold).astype(int)

            y_true.extend(y)
            y_pred.extend(preds)
            y_prob.extend(probs)

    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred),
        "recall": recall_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred),
        "roc_auc": roc_auc_score(y_true, y_prob),
        "confusion_matrix": confusion_matrix(y_true, y_pred)
    }

    return metrics
```

```python
import tenseal as ts
import numpy as np
import torch
import gc

# ── HE Parameters (keep exactly as you have) ──────────────────────────
HE_POLY_MODULUS  = 16384
HE_SCALE_BITS    = 40
HE_COEFF_MOD_BITS = [60, 40, 40, 40, 40, 60]

SELECTED_LAYERS = [
    "lstm.weight_ih_l0",
    "lstm.weight_hh_l0",
    "fc.weight",
    "fc.bias",
]

# ── RECESS Parameters ─────────────────────────────────────────────────
RECESS_DETECTION_INTERVAL = 5   # run detection every N rounds
RECESS_DIRECTION_THRESHOLD = 0.9510   # cos(18°), from paper
RECESS_TRUST_DECAY = 0.9        # weight decay for older rounds
RECESS_FLAG_THRESHOLD = 0.3     # trust score below this → flagged

# ── Threshold SS Parameters ───────────────────────────────────────────
THRESHOLD_T = 2   # minimum shares needed for reconstruction (2-of-3)

```

```python
def create_ckks_context():

    context = ts.context(
        ts.SCHEME_TYPE.CKKS,
        poly_modulus_degree=HE_POLY_MODULUS,
        coeff_mod_bit_sizes=HE_COEFF_MOD_BITS
    )
    context.global_scale = 2 ** HE_SCALE_BITS
    context.generate_galois_keys()
    context.generate_relin_keys()
    return context


def compute_model_update(local_model, global_model):

    delta = {}
    local_state  = local_model.state_dict()
    global_state = global_model.state_dict()
    for key in SELECTED_LAYERS:
        if key in local_state and key in global_state:
            diff = local_state[key] - global_state[key]
            diff = torch.clamp(diff, min=-10.0, max=10.0)
            delta[key] = diff
    return delta


def encrypt_update(delta, context):

    encrypted = {}
    shapes     = {}
    for key, tensor in delta.items():
        shapes[key] = tensor.shape
        flat = tensor.cpu().detach().numpy().flatten()
        flat = np.nan_to_num(flat, nan=0.0, posinf=0.0, neginf=0.0)
        flat = np.clip(flat, -10.0, 10.0)
        encrypted[key] = ts.ckks_vector(context, flat.tolist())
    return encrypted, shapes


def encrypted_weighted_sum(encrypted_updates, trust_weights):


    result = {}
    clients_list = list(trust_weights.keys())
    total_weight  = sum(trust_weights.values())

    for key in encrypted_updates[0].keys():
        # first client
        w0 = trust_weights[clients_list[0]] / total_weight
        result[key] = encrypted_updates[0][key] * w0
        # remaining clients
        for i in range(1, len(encrypted_updates)):
            wi = trust_weights[clients_list[i]] / total_weight
            result[key] = result[key] + (encrypted_updates[i][key] * wi)
    return result


def decrypt_update(enc_update, shapes, context_with_sk):

    decrypted = {}
    for key, enc_vec in enc_update.items():
        # Serialize the ciphertext and deserialize under the sk context
        cipher_bytes = enc_vec.serialize()
        linked_vec   = ts.ckks_vector_from(context_with_sk, cipher_bytes)
        flat = linked_vec.decrypt()
        flat = np.array(flat, dtype=np.float32)
        flat = np.nan_to_num(flat, nan=0.0, posinf=0.0, neginf=0.0)
        shape = shapes[key]
        flat  = flat[:int(np.prod(shape))]
        decrypted[key] = torch.tensor(flat.reshape(shape), dtype=torch.float32)
    return decrypted
```

```python
def setup_threshold_keys(clients):


    print("  [SETUP] Generating CKKS context with secret key...")
    full_ctx = create_ckks_context()

    # Serialize WITH secret key
    full_bytes = full_ctx.serialize(save_secret_key=True)
    full_arr   = np.frombuffer(full_bytes, dtype=np.uint8).copy()
    n_bytes    = len(full_arr)




    sk_shares = {}
    running_sum = np.zeros(n_bytes, dtype=np.int32)
    for i, client in enumerate(clients[:-1]):
        share = np.random.randint(0, 256, size=n_bytes, dtype=np.uint8)
        sk_shares[client] = share
        running_sum = running_sum + share.astype(np.int32)

    last_client = clients[-1]
    sk_shares[last_client] = ((full_arr.astype(np.int32) - running_sum) % 256).astype(np.uint8)



    # Verify reconstruction (sanity check at setup time only)
    reconstructed = np.zeros(n_bytes, dtype=np.int32)
    for share in sk_shares.values():
        reconstructed = reconstructed + share.astype(np.int32)
    reconstructed_mod = (reconstructed % 256).astype(np.uint8)
    assert np.array_equal(reconstructed_mod, full_arr), "Share splitting failed!"

    # Make public context (no secret key) — this is what PS and clients use for encryption
    public_ctx = full_ctx.copy()
    public_ctx.make_context_public()


    del full_ctx
    del full_arr
    del full_bytes
    gc.collect()
    print("  [SETUP]  Full secret key deleted. Public context ready.")
    print("  [SETUP] Each client now holds ONLY their own share.")

    return public_ctx, sk_shares, n_bytes


def threshold_reconstruct_and_decrypt(enc_update, shapes, sk_shares,
                                       participating_clients, n_bytes):

    assert len(participating_clients) >= THRESHOLD_T, \
        f"Need at least {THRESHOLD_T} clients for threshold decryption, got {len(participating_clients)}"

    print(f"  [THRESHOLD] Reconstructing sk from shares: {participating_clients}")

    # Step 1+2: Sum participating shares
    reconstructed = np.zeros(n_bytes, dtype=np.uint8)
    for client in participating_clients:
        share = sk_shares[client]
        reconstructed = (reconstructed.astype(np.int16) + share.astype(np.int16)).astype(np.uint8)



    # Step 3: Deserialize into temporary context
    temp_ctx = ts.context_from(bytes(reconstructed.tobytes()))

    # Step 4: Decrypt
    result = decrypt_update(enc_update, shapes, temp_ctx)

    # Step 5: Destroy temporary context and reconstructed key
    del temp_ctx
    del reconstructed
    gc.collect()
    print("  [THRESHOLD]  Decryption complete. Temporary key destroyed.")

    return result
```

```python
#PART C — RECESS DETECTION  (new)
# ════════════════════════════════════════════════════════════════════════

# PS-side state — initialized before training loop
trust_scores      = {}   # {client_name: float}  — initialized to 1.0
trust_history     = {}   # {client_name: [list of per-round abnormality scores]}
flagged_clients   = set()
last_agg_gradient = {}   # {layer_key: np.ndarray} — previous round's plaintext agg gradient
gradient_archive  = {}   # {client_name: {round_idx: enc_delta}} — for FedRecovery


def init_recess_state(clients):
    """Call once before training loop starts."""
    global trust_scores, trust_history, flagged_clients, gradient_archive
    trust_scores    = {c: 1.0 for c in clients}
    trust_history   = {c: []  for c in clients}
    flagged_clients = set()
    gradient_archive = {c: {} for c in clients}
    print("  [RECESS] State initialized for clients:", clients)


def construct_test_gradient(last_gradient, direction_threshold=RECESS_DIRECTION_THRESHOLD):
    """
    RECESS Step 2.2 — Build personalized test gradient from last round's gradient.

    From paper:
      g^t_i = g^(t-1) / ||g^(t-1)||_2   (normalize)
      then add noise to first 10% of dims until cosine similarity meets threshold A.

    Returns a dict {layer_key: np.ndarray} — one test gradient per layer.
    """
    test_grad = {}
    for key, arr in last_gradient.items():
        flat = arr.flatten().astype(np.float64)
        norm = np.linalg.norm(flat)
        if norm < 1e-10:
            test_grad[key] = arr.copy()
            continue

        # Normalize
        g_normalized = flat / norm

        # Perturb first 10% of dimensions iteratively until cosine sim meets threshold
        n_perturb = max(1, int(0.1 * len(flat)))
        perturbed = g_normalized.copy()
        for _ in range(100):   # max 100 tries
            noise = np.random.normal(0, 0.01, n_perturb)
            candidate = g_normalized.copy()
            candidate[:n_perturb] += noise
            cos_sim = np.dot(g_normalized, candidate) / (
                np.linalg.norm(g_normalized) * np.linalg.norm(candidate) + 1e-10
            )
            if cos_sim >= direction_threshold:
                perturbed = candidate
                break

        test_grad[key] = perturbed.reshape(arr.shape).astype(np.float32)
    return test_grad


def simulate_client_response(client_name, test_gradient, client_loader,
                              global_model, lr=1e-3):
    """
    RECESS Step 2.4 — Client receives test gradient and fine-tunes locally.

    In simulation: client trains one step using the test gradient as the
    'received aggregation gradient' signal, then returns updated gradient.

    Returns: response_gradient dict {layer_key: np.ndarray}
    """
    # Clone global model for this client
    local_model = CNN_LSTM_IDS(SEQ_LEN, NUM_FEATURES).to(DEVICE)
    local_model.load_state_dict(global_model.state_dict())

    # Apply test gradient as a model update (simulates client receiving it)
    with torch.no_grad():
        state = local_model.state_dict()
        for key, grad_arr in test_gradient.items():
            if key in state:
                grad_t = torch.tensor(grad_arr, dtype=torch.float32).to(DEVICE)
                state[key] = state[key] + lr * grad_t
        local_model.load_state_dict(state)

    # Client trains ONE batch on their local data
    local_model.train()
    optimizer = torch.optim.Adam(local_model.parameters(), lr=lr)
    for x_batch, y_batch in client_loader:
        x_batch = x_batch.to(DEVICE)
        y_batch = y_batch.to(DEVICE).unsqueeze(1)
        optimizer.zero_grad()
        loss = CRITERION(local_model(x_batch), y_batch)
        loss.backward()
        optimizer.step()
        break  # one batch only

    # Compute response gradient = updated local model - global model
    response = compute_model_update(local_model, global_model)
    response_np = {k: v.cpu().numpy() for k, v in response.items()}

    del local_model
    torch.cuda.empty_cache()
    return response_np


def simulate_malicious_response(test_gradient, global_model):
    """
    Simulates a poisoning client that tries to deviate the gradient direction.
    Used for testing RECESS detection. Returns a deliberately wrong gradient.
    """
    response = {}
    for key, arr in test_gradient.items():
        flat = arr.flatten()
        # Poisoning: reverse direction + amplify
        poisoned = -flat * np.random.uniform(1.5, 3.0)
        response[key] = poisoned.reshape(arr.shape).astype(np.float32)
    return response


def compute_abnormality(test_grad, response_grad):
    """
    RECESS Step 2.6 — Compute abnormality score from direction + magnitude.

    Direction:  cosine similarity between test and response (lower = more abnormal)
    Magnitude:  L2 norm of response (higher = more abnormal for poisoning)

    Returns: (abnormality_score float, cosine_sim float, l2_norm float)
    """
    all_test     = np.concatenate([v.flatten() for v in test_grad.values()])
    all_response = np.concatenate([v.flatten() for v in response_grad.values()])

    # Cosine similarity
    norm_t = np.linalg.norm(all_test)
    norm_r = np.linalg.norm(all_response)

    if norm_t < 1e-10 or norm_r < 1e-10:
        return 1.0, 0.0, norm_r  # treat zero gradient as maximally abnormal

    cos_sim = np.dot(all_test, all_response) / (norm_t * norm_r)
    cos_sim = float(np.clip(cos_sim, -1.0, 1.0))

    # Direction abnormality: 1 - cosine_similarity (0=same dir, 2=opposite)
    direction_score = 1.0 - cos_sim

    # Magnitude score: normalize by test gradient norm
    magnitude_score = norm_r / (norm_t + 1e-10)

    # Combined abnormality (equal weight, tunable)
    abnormality = 0.5 * direction_score + 0.5 * min(magnitude_score, 2.0) / 2.0

    return float(abnormality), cos_sim, float(norm_r)


def update_trust_scores(client_name, abnormality_score):
    """
    RECESS Step 2.7 — Multi-round exponential decay trust scoring.

    Appends new abnormality to history.
    Recomputes trust score as decayed weighted average of 1 - abnormality.
    """
    trust_history[client_name].append(abnormality_score)
    history = trust_history[client_name]

    # Exponential decay: recent rounds weighted more
    weights = [RECESS_TRUST_DECAY ** (len(history) - 1 - i) for i in range(len(history))]
    total_w = sum(weights)

    trust = sum(w * (1.0 - a) for w, a in zip(weights, history)) / total_w
    trust_scores[client_name] = float(np.clip(trust, 0.0, 1.0))


def run_recess_detection(clients, client_loaders, global_model, round_idx):
    """
    RECESS Phase 2 — Full detection round.

    Called from the main training loop when round_idx % RECESS_DETECTION_INTERVAL == 0.
    Uses last_agg_gradient (set by training loop after each normal round).

    Updates trust_scores and flagged_clients in place.
    """
    global flagged_clients

    if not last_agg_gradient:
        print("  [RECESS] Skipping: no aggregated gradient available yet.")
        return

    print(f"\n  {'='*50}")
    print(f"  [RECESS] Detection round at training round {round_idx+1}")
    print(f"  {'='*50}")

    for client in clients:
        if client in flagged_clients:
            print(f"  [RECESS] {client} already flagged, skipping.")
            continue

        # Step 2.2: construct personalized test gradient
        test_grad = construct_test_gradient(last_agg_gradient)

        # Step 2.4: get client response
        # In real deployment: encrypt test_grad, send to client, client returns enc response
        # In simulation: call simulate_client_response directly
        response_grad = simulate_client_response(
            client, test_grad, client_loaders[client], global_model
        )

        # Step 2.6: compute abnormality
        abnormality, cos_sim, l2_norm = compute_abnormality(test_grad, response_grad)

        # Step 2.7: update trust
        update_trust_scores(client, abnormality)

        ts_val = trust_scores[client]
        print(f"  [RECESS] {client:<10} | cosine_sim={cos_sim:.4f} | "
              f"l2_norm={l2_norm:.4f} | abnormality={abnormality:.4f} | "
              f"trust_score={ts_val:.4f}")

        # Step 2.8: flag if below threshold
        if ts_val < RECESS_FLAG_THRESHOLD:
            flagged_clients.add(client)
            print(f"  [RECESS] ⚠️  {client} FLAGGED as potentially malicious! "
                  f"trust={ts_val:.4f}")

    print(f"  [RECESS] Current trust scores: {trust_scores}")
    print(f"  [RECESS] Flagged clients: {flagged_clients}")

print("✅ Cell 49 loaded: HE functions + Threshold SS + RECESS detection")
```

```python
import gc
global last_agg_gradient
last_agg_gradient = {}
# ── PHASE 0: One-time threshold key setup ─────────────────────────────
print("="*60)
print("PHASE 0: Threshold Secret Sharing Setup")
print("="*60)

public_ctx, sk_shares, sk_n_bytes = setup_threshold_keys(CLIENTS)


for client in CLIENTS:
    print(f"  {client} share size: {len(sk_shares[client])} bytes")

# Initialize RECESS state
init_recess_state(CLIENTS)

# ── PHASE 0: Global model and training config ─────────────────────────
global_model = CNN_LSTM_IDS(SEQ_LEN, NUM_FEATURES).to(DEVICE)
init_path    = os.path.join(MODELS_DIR, "cnn_lstm_init.pt")
global_model.load_state_dict(torch.load(init_path, map_location=DEVICE))
print("\n✅ Global model loaded")

ROUNDS          = CONFIG["ROUNDS"]
CHECKPOINT_FREQ = 2
start_round     = load_latest_checkpoint(global_model)
#start_round = 0
# Active clients (remove flagged ones as training progresses)
active_clients = list(CLIENTS)

print(f"\n{'='*60}")
print(f"Starting FedPHE+TSS+RECESS training: {ROUNDS} rounds")
print(f"{'='*60}\n")

# ── MAIN TRAINING LOOP ────────────────────────────────────────────────
for rnd in range(start_round, ROUNDS):
    print(f"\n{'='*60}")
    print(f"ROUND {rnd+1}/{ROUNDS}  |  Active clients: {active_clients}")
    print(f"{'='*60}")

    # ── RECESS detection round (skip normal aggregation this round) ───
    if rnd > 0 and rnd % RECESS_DETECTION_INTERVAL == 0:
        run_recess_detection(active_clients, client_loaders, global_model, rnd)

        # Remove newly flagged clients from active list
        for fc in flagged_clients:
            if fc in active_clients:
                active_clients.remove(fc)
                print(f"  Removed {fc} from active training clients.")

        if not active_clients:
            print("    All clients flagged. Stopping training.")
            break

        # Detection round does not update the model — continue to next round
        continue


    encrypted_updates = []   # list of enc_delta dicts, one per client
    shapes_ref        = {}   # shapes from last client (all identical)

    for client in active_clients:
        print(f"  Training locally on {client}...")

        # Local training
        local_model = CNN_LSTM_IDS(SEQ_LEN, NUM_FEATURES).to(DEVICE)
        local_model.load_state_dict(global_model.state_dict())
        local_train(
            local_model,
            client_loaders[client],
            CONFIG["LOCAL_EPOCHS"],
            CONFIG["LEARNING_RATE"]
        )

        # Compute plaintext delta
        delta = compute_model_update(local_model, global_model)

        # ── Archive encrypted delta for FedRecovery ──────────────────
        enc_delta_archive, shapes_ref = encrypt_update(delta, public_ctx)
        gradient_archive[client][rnd] = enc_delta_archive
        # ─────────────────────────────────────────────────────────────

        # Encrypt for aggregation (same ciphertext, separate object)
        enc_delta, shapes_ref = encrypt_update(delta, public_ctx)
        encrypted_updates.append(enc_delta)

        del local_model
        torch.cuda.empty_cache()


    active_trust = {c: trust_scores[c] for c in active_clients}
    print(f"  Aggregation weights (trust scores): {active_trust}")

    enc_sum = encrypted_weighted_sum(encrypted_updates, active_trust)


    delta_avg = threshold_reconstruct_and_decrypt(
        enc_sum,
        shapes_ref,
        sk_shares,
        participating_clients=list(CLIENTS),   # all active clients contribute shares
        n_bytes=sk_n_bytes
    )

    # Sanity check on round 0
    if rnd == 0:
        norm = sum(torch.norm(v).item() for v in delta_avg.values())
        print(f"  ΔW norm (sanity check round 0): {norm:.6f}")

    # ── Apply decrypted update to global model ────────────────────────
    current_state = global_model.state_dict()
    for k in delta_avg:
        current_state[k] = current_state[k] + delta_avg[k].to(DEVICE)
    global_model.load_state_dict(current_state)

    # ── Store plaintext aggregated gradient for RECESS (next detection) ─
    # This is the gradient RECESS will use to construct test gradients
    last_agg_gradient = {k: v.cpu().numpy() for k, v in delta_avg.items()}

    print(f"  ✅ Global model updated (round {rnd+1})")

    # ── Evaluation ───────────────────────────────────────────────────
    if (rnd + 1) % 5 == 0:
        metrics = evaluate_global_model(global_model, eval_loader, threshold=0.3)
        print(f"  📊 Round {rnd+1} Eval | "
              f"F1={metrics['f1']:.4f} | "
              f"AUC={metrics['roc_auc']:.4f} | "
              f"Acc={metrics['accuracy']:.4f}")

    # ── Checkpoint ───────────────────────────────────────────────────
    if (rnd + 1) % CHECKPOINT_FREQ == 0:
        save_checkpoint(rnd, global_model)

    gc.collect()

print("\n" + "="*60)
print("Training complete.")
print(f"Flagged clients: {flagged_clients}")
print(f"Active clients at end: {active_clients}")
print(f"Trust scores: {trust_scores}")
print("="*60)
```
